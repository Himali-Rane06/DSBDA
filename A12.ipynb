```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load iris dataset
df = sns.load_dataset('iris')

# Display dataset
print(df.head())

# Dataset info
print(df.info())

# Data types
print(df.dtypes)

# ---------------- HISTOGRAMS ----------------

plt.hist(df['sepal_length'], bins=20)
plt.title("Sepal Length")
plt.show()

plt.hist(df['sepal_width'], bins=20)
plt.title("Sepal Width")
plt.show()

plt.hist(df['petal_length'], bins=20)
plt.title("Petal Length")
plt.show()

plt.hist(df['petal_width'], bins=20)
plt.title("Petal Width")
plt.show()

# ---------------- BOXPLOTS ----------------

sns.boxplot(y=df['sepal_length'])
plt.show()

sns.boxplot(y=df['sepal_width'])
plt.show()

sns.boxplot(y=df['petal_length'])
plt.show()

sns.boxplot(y=df['petal_width'])
plt.show()

# ---------------- SUBPLOTS ----------------

fig, axes = plt.subplots(2, 2, figsize=(10,8))

axes[0,0].hist(df['sepal_length'])
axes[0,0].set_title("Sepal Length")

axes[0,1].hist(df['sepal_width'])
axes[0,1].set_title("Sepal Width")

axes[1,0].hist(df['petal_length'])
axes[1,0].set_title("Petal Length")

axes[1,1].hist(df['petal_width'])
axes[1,1].set_title("Petal Width")

plt.tight_layout()

plt.show()
```
