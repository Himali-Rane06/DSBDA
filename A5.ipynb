```python
# Import libraries
import pandas as pd
import numpy as np

# ---------------- TITANIC DATASET ----------------

# Load Titanic dataset
df = pd.read_csv("C:/Users/rudra/Downloads/Titanic-Dataset.csv")

# Summary statistics grouped by Sex
grouped_stats = df.groupby('Sex')['Fare'].agg([
    'mean',
    'median',
    'min',
    'max',
    'std'
])

print("\nSummary Statistics")
print(grouped_stats)

# Convert categorical variable to numeric
df['Sex_numeric'] = df['Sex'].map({
    'male': 0,
    'female': 1
})

print("\nCategorical to Numeric")
print(df[['Sex', 'Sex_numeric']].head())

# ---------------- IRIS DATASET ----------------

# Load dataset
iris = pd.read_csv("C:/Users/rudra/Downloads/13Iris.csv")

# Display columns
print(iris.columns)

# Group by species
grouped = iris.groupby('species')

# Statistical details
for name, group in grouped:

    print("\nSpecies:", name)

    print("\nMean")
    print(group.mean(numeric_only=True))

    print("\nMedian")
    print(group.median(numeric_only=True))

    print("\nStandard Deviation")
    print(group.std(numeric_only=True))

    print("\nPercentiles")
    print(group.quantile([0.25, 0.50, 0.75], numeric_only=True))
```
